## In- Lab Task:


### Algorithm 
##### Step 1: Import the required Python libraries. 
##### Step 2: Load the pre-trained GloVe model. 
##### Step 3: Define two input sentences. 
##### Step 4: Convert each word into its GloVe embedding. 
##### Step 5: Generate the sentence embedding by averaging the word vectors. 
##### Step 6: Compute cosine similarity between the two sentence embeddings. 
##### Step 7: Display the similarity score. 

### Sample Input 
##### Sentence 1: I love machine learning 
##### Sentence 2: I enjoy artificial intelligence 
### Expected Output 
##### Sentence 1 : I love machine learning 
##### Sentence 2 : I enjoy artificial intelligence 
##### Cosine Similarity : 0.82

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
import gensim.downloader as api

# Load pre-trained GloVe model
glove_model = api.load("glove-wiki-gigaword-100")

# Input sentences
sentence1 = "I love machine learning"
sentence2 = "I enjoy artificial intelligence"

# Function to create sentence embedding
def sentence_embedding(sentence):
    words = sentence.lower().split()
    vectors = [glove_model[word] for word in words if word in glove_model]

    if len(vectors) == 0:
        return np.zeros(glove_model.vector_size)

    return np.mean(vectors, axis=0)

# Generate sentence embeddings
embedding1 = sentence_embedding(sentence1)
embedding2 = sentence_embedding(sentence2)

# Calculate cosine similarity
similarity = cosine_similarity(
    embedding1.reshape(1, -1),
    embedding2.reshape(1, -1)
)[0][0]

# Display output
print("Sentence 1 :", sentence1)
print("Sentence 2 :", sentence2)
print("Cosine Similarity :", round(similarity, 2))
```

```python

from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity

# Corpus
sentences = [
    ["i", "love", "machine", "learning"],
    ["i", "enjoy", "artificial", "intelligence"],
    ["cats", "drink", "milk"],
    ["dogs", "drink", "milk"]
]

# Train Word2Vec
model = Word2Vec(
    sentences,
    vector_size=50,
    window=2,
    min_count=1,
    sg=1
)

# Sentence embedding function
def sentence_vector(sentence):
    words = sentence.lower().split()
    vectors = [model.wv[word] for word in words]
    return sum(vectors) / len(vectors)

# Two sentences
s1 = "i love machine learning"
s2 = "i enjoy artificial intelligence"

# Convert sentences to vectors
v1 = sentence_vector(s1)
v2 = sentence_vector(s2)

# Calculate similarity
similarity = cosine_similarity([v1], [v2])[0][0]

print("Sentence 1:", s1)
print("Sentence 2:", s2)
print("Cosine Similarity:", similarity)
```

## Post-Lab Task:

###   1. What is GloVe?

##### A) GloVe stands for Global Vectors for Word Representation. It is a word embedding technique that represents words as numerical vectors based on their relationships and co-occurrence in a large text corpus.

###   2. What is meant by "Global Vectors"?

##### A) "Global Vectors" means that GloVe uses global word co-occurrence information from the entire corpus to learn relationships between words.

###   3. How is GloVe different from Word2Vec?

##### A) GloVe: Uses global word co-occurrence statistics. Word2Vec: Mainly learns word relationships by predicting words from their local context. Both methods produce numerical vector representations of words.

###   4. Why is GloVe called a pre-trained embedding model?

##### A) GloVe is called a pre-trained embedding model because its word vectors are already learned from a large text corpus. These vectors can be directly used in NLP applications without training the model from scratch.
