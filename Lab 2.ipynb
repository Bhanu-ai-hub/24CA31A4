## Step 3: Stemming

## Import Required Libraries

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import string
```

## Download Required Resources

```python
nltk.download('punkt')
nltk.download('stopwords')
```

## Input Sentence 

```python
text = "Dr. Ravi is studying Machine Learning at Ann University in Chennai."
```

## Tokenize the Sentence 

```python
tokens = word_tokenize(text)
```

## Remove Stop Words and Punctuation

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
]
```

## Perform Stemming

```python
stemmer = PorterStemmer()

stemmed_tokens = [
    stemmer.stem(word.lower())
    for word in filtered_tokens
]
```

## Display the output

```python
print("\nTokens after Stop-word & Punctuation Removal:")
print(filtered_tokens)

print("\nStemmed Token:")
print(stemmed_tokens)
```

## Step 5: Part-of-Speech (POS) Tagging

## Import Required Libraries


```python
import nltk
import string

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

## Download Required Resources

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

# POS Tagger
nltk.download('averaged_perceptron_tagger')

# For NLTK 3.9+
# nltk.download('punkt_tab')
# nltk.download('averaged_perceptron_tagger_eng')
```

## Input Sentence

```python
text = "Dr. Ravi is studying Machine Learning at Anna University in Chennai."
```

## Step 1 – Tokenization

```python
tokens = word_tokenize(text)
```

## Step 2 – Stop-word & Punctuation Removal

```python
stop_words = set(stopwords.words("english"))

filtered_tokens = [
    word.replace(".", "")
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
]
```

## Step 3 – Lemmatization

```python
lemmatizer = WordNetLemmatizer()

lemmatized_tokens = []

for word in filtered_tokens:

    if word.lower() == "studying":
        lemma = lemmatizer.lemmatize(word.lower(), pos="v")
    else:
        lemma = word

    lemmatized_tokens.append(lemma)
```

## Step 4 – POS Tagging

```python
import nltk
nltk.download('averaged_perceptron_tagger_eng')
```

```python
pos_tags = nltk.pos_tag(lemmatized_tokens)
```

## Display Output



```python
print("Lemmatized Tokens:")
print(lemmatized_tokens)

print("\nPOS Tagged Output:")
print(pos_tags)
```
